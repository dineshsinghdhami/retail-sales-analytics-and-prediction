## Step 4: Data Cleaning & Preprocessing

### Objective

Before performing data analysis, visualization, SQL analysis, and machine learning, the dataset must be cleaned and prepared.

In this notebook, we will:

- Load the original dataset
- Check for missing values
- Handle missing values
- Check for duplicate records
- Convert date columns into proper datetime format
- Check numerical values for possible problems
- Create useful date-related features
- Save the cleaned dataset for future analysis

## 1. Import Required Libraries

We will use Pandas for data manipulation and preprocessing.

```python
import pandas as pd

print("Pandas imported successfully!")
```

## 2. Load the Original Dataset

We will load the original `train.csv` file into a Pandas DataFrame named `df`.

```python
df = pd.read_csv("train.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 3. Check Missing Values

Missing values are records where information is unavailable.

We will count the missing values in each column before deciding how to handle them.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

### Missing Value Observation

The initial inspection shows that the dataset contains a small number of missing values in the `Postal Code` column.

Postal Code represents geographic information rather than the main sales measurement. Instead of deleting complete sales records because of a missing postal code, we will fill the missing values with `0`.

This allows us to preserve the sales records for further analysis.

```python
df["Postal Code"] = df["Postal Code"].fillna(0)

print("Missing values after handling Postal Code:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

## 4. Check for Duplicate Records

Duplicate records can cause incorrect calculations because the same observation may be counted more than once.

We will check whether the dataset contains completely duplicated rows.

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

### Remove Duplicate Records

If duplicate rows exist, they will be removed to prevent them from affecting the analysis.

```python
rows_before = len(df)

df = df.drop_duplicates()

rows_after = len(df)

print("Rows before removing duplicates:", rows_before)
print("Rows after removing duplicates:", rows_after)
print("Duplicates removed:", rows_before - rows_after)
```

## 5. Convert Date Columns

The `Order Date` and `Ship Date` columns were initially stored as text (`object`).

For time-based sales analysis, these columns need to be converted into proper datetime values.

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d/%m/%Y"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%d/%m/%Y"
)

print("Date columns converted successfully!")

print("\nData types:")
print(df[["Order Date", "Ship Date"]].dtypes)
```

## 6. Check the Sales Column

Sales is the main numerical variable in this project.

We will inspect its basic statistics and check whether the dataset contains zero or negative sales values.

```python
print("Sales summary:")
print(df["Sales"].describe())

print("\nNumber of zero sales values:")
print((df["Sales"] == 0).sum())

print("\nNumber of negative sales values:")
print((df["Sales"] < 0).sum())
```

## 7. Create Date-Based Features

To make time-based analysis easier, we will extract additional information from the `Order Date` column.

The following new columns will be created:

- `Order Year`
- `Order Month`
- `Order Month Name`
- `Order Quarter`

These features will help us analyze sales trends over time.

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month Name"] = df["Order Date"].dt.month_name()
df["Order Quarter"] = df["Order Date"].dt.quarter

print("Date features created successfully!")

df[
    [
        "Order Date",
        "Order Year",
        "Order Month",
        "Order Month Name",
        "Order Quarter"
    ]
].head()
```

## 8. Calculate Shipping Duration

Shipping duration represents the number of days between the order date and the shipping date.

This can help us understand how quickly orders are shipped.

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

print("Shipping Days summary:")
print(df["Shipping Days"].describe())
```

## 9. Verify the Cleaned Dataset

After completing the cleaning process, we will verify:

- Dataset dimensions
- Missing values
- Duplicate records
- Data types
- Newly created columns

```python
print("=" * 50)
print("CLEANED DATASET INFORMATION")
print("=" * 50)

print("\nRows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

## 10. Save the Cleaned Dataset

The cleaned and preprocessed dataset will be saved as `cleaned_superstore.csv`.

This cleaned dataset will be used in the next stages of the project, including:

- Exploratory Data Analysis (EDA)
- Data visualization
- SQL analysis
- Dashboard development
- Machine learning

```python
df.to_csv("cleaned_superstore.csv", index=False)

print("Cleaned dataset saved as cleaned_superstore.csv")
```

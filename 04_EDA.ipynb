## 1. Import Required Libraries

We will use:

- **Pandas** for data analysis
- **NumPy** for numerical operations
- **Matplotlib** for data visualization
- **Seaborn** for statistical visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Required libraries imported successfully!")
```

## 2. Load the Cleaned Dataset

We will use the cleaned dataset created during the data preprocessing stage.

```python
df = pd.read_csv("cleaned_superstore.csv")

print("Cleaned dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

### Convert Date Columns

Since CSV files do not preserve Pandas datetime data types, the order and shipping date columns are converted back into datetime format before performing time-based analysis.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print("Order Date type:", df["Order Date"].dtype)
print("Ship Date type:", df["Ship Date"].dtype)

print("\n Date columns ready for analysis!")
```

## 3. Key Performance Indicators (KPIs)

Key Performance Indicators provide a quick overview of the overall performance of the retail business.

We will calculate:

1. Total Sales
2. Total Orders
3. Total Customers
4. Average Order Value

```python
total_sales = df["Sales"].sum()

total_orders = df["Order ID"].nunique()

total_customers = df["Customer ID"].nunique()

average_order_value = total_sales / total_orders

print("=" * 50)
print("RETAIL SALES - KEY PERFORMANCE INDICATORS")
print("=" * 50)

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
```

## 4. Sales Analysis by Product Category

### Business Question

**Which product category generates the highest sales?**

Understanding category-level performance helps a retail business identify which groups of products contribute the most to overall revenue.

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Category:")
print(category_sales)
```

### Visualization: Sales by Category

A bar chart makes it easier to compare total sales across the different product categories.

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales ($)")

plt.tight_layout()
plt.show()
```

### Business Insight

The category-level analysis shows that **Technology generates the highest total sales**, followed by **Furniture** and **Office Supplies**.

This indicates that Technology products are a major contributor to the company's overall sales revenue. The business may consider focusing marketing campaigns, inventory planning, and promotional strategies around high-performing Technology products while also investigating opportunities to improve sales in the lower-performing categories.

## 5. Sales Analysis by Region

Regional sales analysis helps identify strong and weak geographical markets and can support decisions related to marketing, inventory distribution, and expansion.

### Business Insight

The regional analysis shows that the **West region generates the highest total sales**.

This suggests that the West is an important market for the business. The company can study the products and customer behavior driving strong performance in this region and potentially apply similar strategies to lower-performing regions.

## 6. Sales Analysis by Customer Segment

Customer segmentation analysis helps the business understand which types of customers generate the most revenue.

## 10. Yearly Sales Trend

### Business Question

**How have total sales changed from year to year?**

Analyzing yearly sales helps us understand the long-term direction of business performance and identify whether revenue is generally increasing or decreasing.

```python
yearly_sales = (
    df.groupby("Order Year")["Sales"]
    .sum()
    .sort_index()
)

print("Yearly Sales:")
print(yearly_sales)
```

### Visualization: Yearly Sales Trend

A line chart is useful for showing how sales change over time.

```python
plt.figure(figsize=(9, 5))

plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker="o"
)

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales ($)")
plt.xticks(yearly_sales.index)
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Business Insight

The yearly sales trend provides an overview of how the company's revenue has changed over time.

The visualization indicates an overall growth pattern across the available years, although the amount of growth may vary from one year to another.

This historical trend is particularly important because past sales patterns can later be used as part of the sales prediction process.

## 11. Monthly Sales Trend

### Business Question

**How do sales change from month to month over the complete historical period?**

Monthly analysis provides a more detailed view of sales performance and can reveal growth patterns, seasonal changes, and unusual increases or decreases in revenue.

```python
monthly_sales = (
    df.set_index("Order Date")
    .resample("ME")["Sales"]
    .sum()
)

print("Monthly Sales:")
print(monthly_sales.head(12))
```

### Visualization: Monthly Sales Trend

The following line chart displays total monthly sales across the complete historical period.

```python
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    markersize=4
)

plt.title("Monthly Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Total Sales ($)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Business Insight

The monthly sales trend shows noticeable fluctuations in revenue across different months.

Some periods experience significantly higher sales than others, suggesting that customer demand may vary throughout the year. These patterns may be influenced by seasonal demand, promotional periods, customer purchasing behavior, or other business factors.

Understanding monthly sales patterns can support inventory planning, marketing decisions, and future sales forecasting.

## 12. Shipping Performance Analysis

### Business Question

**How does average shipping duration differ between shipping modes?**

Shipping performance is important for customer satisfaction and operational efficiency. Comparing shipping duration across shipping modes helps us understand how quickly different delivery options operate.

```python
shipping_analysis = (
    df.groupby("Ship Mode")["Shipping Days"]
    .mean()
    .sort_values()
)

print("Average Shipping Days by Ship Mode:")
print(shipping_analysis)
```

### Visualization: Average Shipping Days by Ship Mode

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=shipping_analysis.index,
    y=shipping_analysis.values
)

plt.title("Average Shipping Days by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Average Shipping Days")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

### Business Insight

The shipping analysis shows differences in average delivery time across the available shipping modes.

Faster shipping options generally require fewer days between the order date and ship date, while standard delivery methods take longer.

This information can help the business evaluate delivery performance and understand the operational differences between shipping options.

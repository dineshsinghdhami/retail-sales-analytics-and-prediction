SQL is widely used by data analysts to retrieve, filter, aggregate, and analyze data stored in databases.

## 1. Import Required Libraries

We will use:

- **Pandas** to load and display data
- **SQLite3** to create and query a SQL database

SQLite3 is included with Python, so no additional database installation is required.

```python
import pandas as pd
import sqlite3

print("Required libraries imported successfully!")
```

## 2. Load the Cleaned Dataset

We will load the cleaned Superstore dataset created during the data cleaning stage.

```python
df = pd.read_csv("cleaned_superstore.csv")

print("Cleaned dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 3. Create the SQLite Database

A database is an organized collection of data.

We will create a SQLite database named `retail_sales.db`. This database will store our cleaned retail sales data and allow us to analyze it using SQL queries.

```python
conn = sqlite3.connect("retail_sales.db")

print("SQLite database created successfully!")
```

## 4. Create the Sales Table

The cleaned Pandas DataFrame will be stored inside the SQLite database as a table named `sales`.

This allows us to query the retail dataset using SQL commands.

```python
df.to_sql(
    "sales",
    conn,
    if_exists="replace",
    index=False
)

print("SQL table 'sales' created successfully!")
```

## 5. View Sales Records Using SQL

### SQL Concept: SELECT

The `SELECT` statement is used to retrieve data from a database table.

The following query retrieves the first five records from the `sales` table.

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

result = pd.read_sql_query(query, conn)

result
```

## 6. Select Specific Columns

Instead of retrieving every column, SQL can be used to select only the columns required for analysis.

The following query retrieves order, customer, category, region, and sales information.

```python
query = """
SELECT
    "Order ID",
    "Customer Name",
    "Category",
    "Region",
    "Sales"
FROM sales
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

result
```

## 7. Filter Records Using WHERE

### Business Question

**Which sales records belong to the Technology category?**

The SQL `WHERE` clause is used to filter records based on a condition.

```python
query = """
SELECT
    "Order ID",
    "Product Name",
    "Category",
    "Sales"
FROM sales
WHERE "Category" = 'Technology'
LIMIT 10;
"""

technology_sales = pd.read_sql_query(query, conn)

technology_sales
```

## 8. Calculate Total Sales

### Business Question

**What is the total sales revenue generated by the business?**

The SQL `SUM()` function is used to calculate the total of a numerical column.

```python
query = """
SELECT
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales;
"""

total_sales_sql = pd.read_sql_query(query, conn)

total_sales_sql
```

## 9. Count Total Orders

### Business Question

**How many unique orders are present in the dataset?**

Since one order can contain multiple products, we use `COUNT(DISTINCT ...)` to count each Order ID only once.

```python
query = """
SELECT
    COUNT(DISTINCT "Order ID") AS "Total Orders"
FROM sales;
"""

total_orders_sql = pd.read_sql_query(query, conn)

total_orders_sql
```

## 10. Count Total Customers

### Business Question

**How many unique customers purchased products from the business?**

We will count distinct Customer IDs to determine the number of individual customers.

```python
query = """
SELECT
    COUNT(DISTINCT "Customer ID") AS "Total Customers"
FROM sales;
"""

total_customers_sql = pd.read_sql_query(query, conn)

total_customers_sql
```

## 11. Identify High-Value Sales Records

### Business Question

**Which sales records have a sales value greater than $1,000?**

High-value transactions can be useful for identifying important products, customers, and purchasing patterns.

```python
query = """
SELECT
    "Order ID",
    "Customer Name",
    "Product Name",
    "Category",
    "Sales"
FROM sales
WHERE "Sales" > 1000
ORDER BY "Sales" DESC
LIMIT 10;
"""

high_value_sales = pd.read_sql_query(query, conn)

high_value_sales
```

## 12. SQL Database Verification

Before performing more advanced SQL business analysis, we will verify that the database and sales table are working correctly.

```python
query = """
SELECT COUNT(*) AS "Total Rows"
FROM sales;
"""

row_check = pd.read_sql_query(query, conn)

print("=" * 55)
print("SQL DATABASE VERIFICATION")
print("=" * 55)

print("\nRows stored in SQL:")
print(row_check)

print("\nDatabase file: retail_sales.db")
print("Table name: sales")

print("\n SQL DATABASE IS WORKING SUCCESSFULLY!")
```

## 14. Sales by Category

### Business Question

**Which product category generates the highest total sales?**

We will use `GROUP BY` to combine records belonging to the same category and calculate total sales for each category.

```python
query = """
SELECT
    "Category",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Category"
ORDER BY "Total Sales" DESC;
"""

category_sql = pd.read_sql_query(query, conn)

category_sql
```

## 15. Sales by Region

### Business Question

**Which geographical region generates the highest total sales?**

Regional analysis helps identify the strongest and weakest markets.

```python
query = """
SELECT
    "Region",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Region"
ORDER BY "Total Sales" DESC;
"""

region_sql = pd.read_sql_query(query, conn)

region_sql
```

## 16. Sales by Customer Segment

### Business Question

**Which customer segment contributes the highest sales?**

Customer segment analysis helps the business understand which types of customers generate the most revenue.

```python
query = """
SELECT
    "Segment",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Segment"
ORDER BY "Total Sales" DESC;
"""

segment_sql = pd.read_sql_query(query, conn)

segment_sql
```

## 17. Sales by Sub-Category

### Business Question

**Which product sub-categories generate the highest sales?**

Sub-category analysis provides more detailed information about product performance.

```python
query = """
SELECT
    "Sub-Category",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Sub-Category"
ORDER BY "Total Sales" DESC;
"""

subcategory_sql = pd.read_sql_query(query, conn)

subcategory_sql
```

## 18. Top 10 Products by Sales

### Business Question

**Which individual products generate the highest total sales?**

This analysis helps identify the most valuable products in the retail business.

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Product Name"
ORDER BY "Total Sales" DESC
LIMIT 10;
"""

top_products_sql = pd.read_sql_query(query, conn)

top_products_sql
```

## 19. Top 10 Customers by Sales

### Business Question

**Which customers contribute the highest amount of sales?**

Identifying high-value customers can support loyalty, retention, and personalized marketing strategies.

```python
query = """
SELECT
    "Customer Name",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Customer Name"
ORDER BY "Total Sales" DESC
LIMIT 10;
"""

top_customers_sql = pd.read_sql_query(query, conn)

top_customers_sql
```

## 20. Yearly Sales Analysis

### Business Question

**How have total sales changed across different years?**

Yearly sales analysis helps identify long-term growth or decline in business performance.

```python
query = """
SELECT
    "Order Year",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY "Order Year"
ORDER BY "Order Year";
"""

yearly_sales_sql = pd.read_sql_query(query, conn)

yearly_sales_sql
```

## 21. Monthly Sales Analysis

### Business Question

**Which months generate the highest total sales?**

Monthly analysis can help identify seasonal patterns and periods of stronger customer demand.

```python
query = """
SELECT
    "Order Month",
    "Order Month Name",
    ROUND(SUM("Sales"), 2) AS "Total Sales"
FROM sales
GROUP BY
    "Order Month",
    "Order Month Name"
ORDER BY "Order Month";
"""

monthly_sales_sql = pd.read_sql_query(query, conn)

monthly_sales_sql
```

## 22. Number of Orders by Region

### Business Question

**Which region receives the highest number of unique orders?**

Sales value alone does not show how frequently customers place orders. Therefore, we will also compare the number of unique orders across regions.

```python
query = """
SELECT
    "Region",
    COUNT(DISTINCT "Order ID") AS "Total Orders"
FROM sales
GROUP BY "Region"
ORDER BY "Total Orders" DESC;
"""

orders_by_region = pd.read_sql_query(query, conn)

orders_by_region
```

## 23. Average Sales by Category

### Business Question

**What is the average sales value for each product category?**

The SQL `AVG()` function calculates the average value of a numerical column.

```python
query = """
SELECT
    "Category",
    ROUND(AVG("Sales"), 2) AS "Average Sales"
FROM sales
GROUP BY "Category"
ORDER BY "Average Sales" DESC;
"""

average_category_sales = pd.read_sql_query(query, conn)

average_category_sales
```

## 24. Maximum Sale by Category

### Business Question

**What is the largest individual sales value recorded in each category?**

The SQL `MAX()` function identifies the highest value in a column.

```python
query = """
SELECT
    "Category",
    ROUND(MAX("Sales"), 2) AS "Maximum Sale"
FROM sales
GROUP BY "Category"
ORDER BY "Maximum Sale" DESC;
"""

maximum_sales_sql = pd.read_sql_query(query, conn)

maximum_sales_sql
```

## 25. Average Shipping Time by Ship Mode

### Business Question

**How many days does each shipping mode take on average?**

This analysis helps compare the delivery speed of different shipping options.

```python
query = """
SELECT
    "Ship Mode",
    ROUND(AVG("Shipping Days"), 2) AS "Average Shipping Days"
FROM sales
GROUP BY "Ship Mode"
ORDER BY "Average Shipping Days";
"""

shipping_sql = pd.read_sql_query(query, conn)

shipping_sql
```

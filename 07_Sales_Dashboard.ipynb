# Retail Sales Analytics Dashboard

## Step 8: Interactive-Style Sales Dashboard

### Dashboard Objective

This dashboard provides a high-level overview of retail business performance using the cleaned Superstore Sales dataset.

The dashboard focuses on:

- Total Sales
- Total Orders
- Total Customers
- Average Order Value
- Sales by Category
- Sales by Region
- Sales by Customer Segment
- Monthly Sales Trend
- Top Products
- Top Customers

The objective is to present important business insights in a clear and visual format.

## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Dashboard libraries imported successfully!")
```

## 2. Load the Cleaned Dataset

The cleaned Superstore dataset will be used to create the sales dashboard.

```python
df = pd.read_csv("cleaned_superstore.csv")

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

## 3. Key Performance Indicators

The dashboard begins with four important retail KPIs:

- Total Sales
- Total Orders
- Total Customers
- Average Order Value

```python
total_sales = df["Sales"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
average_order_value = total_sales / total_orders

print("=" * 65)
print("                 RETAIL SALES DASHBOARD")
print("=" * 65)

print(f"\nTOTAL SALES        : ${total_sales:,.2f}")
print(f"TOTAL ORDERS       : {total_orders:,}")
print(f"TOTAL CUSTOMERS    : {total_customers:,}")
print(f"AVERAGE ORDER VALUE: ${average_order_value:,.2f}")

print("\n" + "=" * 65)
```

## 4. Sales by Product Category

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales ($)")

plt.tight_layout()
plt.show()
```

## 5. Sales by Region

```python
region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")

plt.tight_layout()
plt.show()
```

## 6. Sales by Customer Segment

```python
segment_sales = (
    df.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=segment_sales.index,
    y=segment_sales.values
)

plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales ($)")

plt.tight_layout()
plt.show()
```

## 7. Monthly Sales Trend

The monthly sales trend helps identify changes in sales performance over time.

```python
monthly_sales = (
    df.set_index("Order Date")
    .resample("ME")["Sales"]
    .sum()
)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    markersize=4
)

plt.title("Monthly Retail Sales Trend")
plt.xlabel("Date")
plt.ylabel("Total Sales ($)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 8. Top 10 Products by Sales

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 7))

sns.barplot(
    x=top_products.values,
    y=top_products.index
)

plt.title("Top 10 Products by Total Sales")
plt.xlabel("Total Sales ($)")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

## 9. Top 10 Customers by Sales

```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_customers.values,
    y=top_customers.index
)

plt.title("Top 10 Customers by Total Sales")
plt.xlabel("Total Sales ($)")
plt.ylabel("Customer")

plt.tight_layout()
plt.show()
```

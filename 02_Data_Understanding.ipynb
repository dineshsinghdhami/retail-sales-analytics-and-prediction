## 1. Import Required Libraries

Pandas is a Python library used for loading, organizing, cleaning, and analyzing structured data.

```python
import pandas as pd

print("Pandas imported successfully!")
```

## 2. Load the Dataset

The dataset is stored in a CSV file named `train.csv`.

We will load the CSV file into a Pandas DataFrame called `df`. A DataFrame is similar to a table in Excel, consisting of rows and columns.

```python
df = pd.read_csv("train.csv")

print("Dataset loaded successfully!")
```

## 3. Preview the Dataset

We will display the first five rows of the dataset to understand its structure and the type of information it contains.

```python
df.head()
```

## 4. Dataset Dimensions

The shape of the dataset tells us the total number of rows and columns.

- **Rows** represent individual sales records.
- **Columns** represent different attributes or features of those sales records.

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 5. Column Names

Next, we will examine all the columns available in the dataset. Understanding the columns helps us determine what types of business questions can be answered using the data.

```python
print("Columns in the dataset:\n")

for column in df.columns:
    print(column)
```

## 6. Dataset Information and Data Types

The `info()` function provides information about:

- Column names

```python
df.info()
```

## 7. Check for Missing Values

Missing values can affect data analysis and machine learning models. Therefore, we need to identify columns containing missing information before performing further analysis.
